 Write a program to calculate the area and perimeter of a rectangle using user input.

````python
length = float(input("Enter length: "))
width = float(input("Enter width: "))

area = length * width
perimeter = 2 * (length + width)

print("Area:", area)
print("Perimeter:", perimeter)
````

Write a program that checks if a number is even or odd using the modulus operator.

````python
number = int(input("Enter a number: "))
if number % 2 == 0:
    print("Even")
else:
    print("Odd")
````

Calculate simple interest: SI = (Principal × Rate × Time) / 100 using input values.

````python
principal = float(input("Enter principal: "))
rate = float(input("Enter rate: "))
time = float(input("Enter time: "))

si = (principal * rate * time) / 100

print("Simple Interest:", si)
````

Write a program that prints all even numbers from 1 to 50 using a for loop.

 Print a star pyramid pattern using nested loops (5 rows).

````python
for i in range(1, 6):
    for j in range(i):
        print("*", end="")
    print()
````

Write a program that counts how many vowels are in a user-entered word.

````python
word = input("Enter a word: ")
count = 0

for letter in word:
    if letter in "aeiou":
        count += 1

print("Number of vowels:", count)
````
